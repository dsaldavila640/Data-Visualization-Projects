```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file_path = r"D:\Data Analytics Certificate - In Process\DAT - 0202 Data Visualization\Deliverable Project\Project 1 - 10\Project 7 - Excel - Data Visualization.xlsx"
oil_data = pd.read_excel(file_path, sheet_name="Spot_Prices")
oil_data.head()
```

```python
oil_data.info()
```

```python
oil_data[["WTI_USD", "Brent_USD"]].describe().round(2)
```

```python
plt.figure(figsize=(12,6))

plt.plot(oil_data["Date"], oil_data["WTI_USD"], label="WTI", color="orange", linewidth=1)

plt.plot(oil_data["Date"], oil_data["Brent_USD"], label="Brent", color="blue", linewidth=1)

plt.title("WTI and Brent Crude Oil Prices (2000-2025)")
plt.xlabel("Year")
plt.ylabel("Crude Oil Price (USD per Barrel)")

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.boxplot([oil_data["WTI_USD"].dropna(), oil_data["Brent_USD"].dropna()], tick_labels=["WTI","Brent"], patch_artist=True)

plt.title("Distribution of WTI and Brent Crude Oil Prices (2000-2025)")
plt.ylabel("Crude Oil Price (USD per Barrel)")
plt.grid(axis="y" , alpha=0.3)

plt.tight_layout()
plt.show()
```

## Interpretation of Oil Price Visualizations (2000-2025)

### Time-Series Analysis

Specifically, the time series graph shows these key events in oil prices. The graph shows prices rising before 2008 and declining sharply during the 2008-2009 financial crisis, indicating the event occurred during the Great Recession for WTI and Brent crude oil. 

Another event occurred between 2014 and 2016, showing another substantial decline. 

During the COVID-19 pandemic in 2020, WTI and Brent crude oil prices declined sharply, especially with WTI experiencing a severe price collapse. 

Oil prices recovered in 2022 during the war when Russia launched a full-scale invasion of Ukraine. 

### Box Plot Analysis

The box plot compares WTI and Brent from 2000 to 2025, showing the interquartile ranges and whiskers for each benchmark. 

The interquartile ranges overlap, indicating similar price distributions, even though their median prices differ; overall, the ranges differ.

WTI median price: $63.05 per barrel.

Brent median price: $65.67 per barrel

Brent has a slight advantage with a higher median price

WTI has shown the box plot regarding having an extreme negative outlier

Both variables show substantial price variability (U.S. Energy Information Administration, n.d.).

### Conclusion

The visualization shows that crude oil prices are volatile and have the most significant price swings between 2000 and 2025.

The findings suggest that global economic conditions and geopolitical events still influence crude oil supply and demand.

However, these visualizations alone cannot establish whether geopolitical events caused changes in crude oil prices.
