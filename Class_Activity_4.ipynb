```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np # Needed for sorting
```

```python
data = {
    'State': ["TX", "OK", "WV", "ND", "CO", "PA", "GOM", "WY", "LA", "OH", "OTHER"],
    'Production_Bcf': [1200, 280, 185, 160, 145, 90, 85, 65, 60, 55, 65]
}
df = pd.DataFrame(data)

# Display the first few rows to confirm the data is loaded
print(df)
```

```python
df.sort_values(by='Production_Bcf', ascending=False, inplace=True)
```

```python
# Create a list of colors based on the State column
colors = ['blue' if state == 'TX' else 'gray' for state in df['State']]
```

```python
plt.figure(figsize=(10, 6))

# 1. Plot the bars
plt.bar(df['State'], df['Production_Bcf'], color=colors)

# 2. Add Titles and Labels
plt.title('Figure 1. Natural Gas Production by State, 2019', fontsize=14, pad=20)
plt.ylabel('Billion Cubic Feet', fontsize=12)
plt.xlabel('') # Remove the x-axis label, keeping only the state abbreviations

# 3. Adjust Y-axis scale (to match the original scale)
plt.ylim(0, 1250)

# 4. Add source text (similar to the original graph's 'Source' line)
plt.figtext(0.5, 0.01, 
            "Source: Author's elaboration with data from the Energy Information Administration", 
            wrap=True, 
            horizontalalignment='center', 
            fontsize=9)

# 5. Display the plot
plt.grid(axis='y', linestyle='--', alpha=0.7) # Add subtle horizontal grid lines
plt.tight_layout(rect=[0, 0.05, 1, 1]) # Adjust layout to fit titles/labels
plt.show()
```
