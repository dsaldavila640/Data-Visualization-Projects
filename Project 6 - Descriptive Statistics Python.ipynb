```python
variables = ["brent_usd",
"wti_usd",
"brent_wti_spread",
"brent_daily_return_pct",
"brent_30d_vol",
"hormuz_risk_premium_pct",
"opec_production_mbd"]

for col in variables:
    print("\n", col)
    print("Median:",
df[col].median())
    print("Mode:",
df[col].mode().tolist())
    print("Range:",df[col].max() - 
df[col].min())
```

```python
summary = pd.DataFrame({
    "Mean": df[variables].mean(),
    "Median": df[variables].median(),
    "Mode": df[variables].mode().iloc[0],
    "Std Dev": df[variables].std(),
    "Minimun": df[variables].min(),
    "Maximum": df[variables].max(),
    "Range": df[variables].max() - df[variables].min()})
summary.round(2)
```
